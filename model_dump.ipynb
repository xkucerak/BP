```python
from pathlib import Path
import pathlib
from softtools.model_manipulator import ModelManipulate
```

```python
def dump_model(source_path: pathlib.Path, destination_path: pathlib.Path):
    if source_path.is_file() and source_path.suffix == ".onnx":  # for single file
        ModelManipulate.dump_model(source_path, destination_path)
    else:  # for directory
        for file in source_path.rglob("*.onnx"):
            ModelManipulate.dump_model(file, destination_path)
```

```python
def load_model(
    source_path: pathlib.Path,
    destination_path: pathlib.Path,
    model_path: pathlib.Path | None = None,
):
    for model in source_path.iterdir():
        for type in model.iterdir():
            for run in type.iterdir():
                ModelManipulate.load_full_model(run, destination_path, by_layer=False)
```

```python
source_path = Path.cwd() / "models"
destination_path = Path.cwd() / "model_dump"
dump_model(source_path, destination_path)
```
