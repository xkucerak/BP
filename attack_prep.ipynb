```python
# packs exported weights to 4MB chunks

from pathlib import Path

MAX_SIZE = 2**20 * 4
```

```python
load_path = Path.cwd() / "model_dump/"

save_path_parent = Path.cwd() / "model_chunks"

last_path = None
act_index = 0
write_buffer = bytes()
save_path = None


for i in sorted(load_path.rglob("*.bin")):
    load_path_temp = i.parent

    if load_path_temp != last_path:
        if len(write_buffer) != 0:
            for rest in range(0, len(write_buffer), MAX_SIZE):
                chunk = write_buffer[rest : rest + MAX_SIZE]
                (save_path / (str(act_index).zfill(3) + ".bin")).write_bytes(chunk)
                act_index += 1

        write_buffer = bytes()
        act_index = 0
        last_path = load_path_temp
        save_path = save_path_parent / i.parent.parent.stem / i.parent.stem
        save_path.mkdir(parents=True, exist_ok=True)

    write_buffer += i.read_bytes()

if len(write_buffer) != 0:
        for rest in range(0, len(write_buffer), MAX_SIZE):
            chunk = write_buffer[rest : rest + MAX_SIZE]
            (save_path / (str(act_index).zfill(3) + ".bin")).write_bytes(chunk)
            act_index += 1
```
