```python
import torch
from torchvision.models import resnet50, resnet34, resnet18, vgg11
```

```python
model = vgg11(pretrained=True)
torch.onnx.export(
    model,
    torch.ones(1,3,224,224),
    "vgg11_fp32.onnx",
    dynamic_axes={
        "input": {0: "batch"},
        "output": {0: "batch"},
    },
    input_names=["input"],
    output_names=["output"],
)

import onnx
from onnxconverter_common import float16

model = onnx.load("vgg11_fp32.onnx")
model_fp16 = float16.convert_float_to_float16(model,keep_io_types=True)
onnx.save(model_fp16, "vgg11_fp16.onnx")
```
